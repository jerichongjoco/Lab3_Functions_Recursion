```python
import functools

LAST_NAME = "ONGJOCO"
SEED_NUM = 2
FAVORITE_ARTIST = "CUP OF JOE"

# Decorator to record the diagnostic process
def audit_logger(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        execution_log.append(f"START: Executing {func.__name__}")
        result = func(*args, **kwargs)
        execution_log.append(f"END: Finished {func.__name__}")
        return result
    return wrapper

execution_log = []

@audit_logger
def generate_equipment_data():
    base_val = len(LAST_NAME) * 10 + SEED_NUM
    data = [base_val, 'ERR', base_val + 15, 'FAULT', base_val - 10, base_val + 30]
    execution_log.append(f"Generated raw data using artist '{FAVORITE_ARTIST}' seed base.")
    return data

@audit_logger
def validate_and_calculate(raw_data):
    valid_results = []
    diagnostic_results = []
    total = 0
    count = 0
    
    for item in raw_data:
        try:
            # Exception handling for invalid inputs/strings
            if not isinstance(item, (int, float)) and not str(item).isdigit():
                raise ValueError(f"Invalid equipment reading: '{item}'")
            val = float(item)
            valid_results.append(f"{val} -> VALID")
            total += val
            count += 1
            
            # Classification logic
            if val < 50:
                diagnostic_results.append(f"{val} -> LOW")
            elif val <= 100:
                diagnostic_results.append(f"{val} -> OPTIMAL")
            else:
                diagnostic_results.append(f"{val} -> CRITICAL")
                
        except ValueError as e:
            execution_log.append(f"WARNING: {e}")
            valid_results.append(f"{item} -> INVALID")
            
    avg = total / count if count > 0 else 0
    return valid_results, diagnostic_results, avg

def run_exercise_one():
    print("=== EXERCISE 1: EQUIPMENT DIAGNOSTIC SYSTEM ===")
    raw_data = generate_equipment_data()
    valid_res, diag_res, avg_val = validate_and_calculate(raw_data)
    
    print("\n--- Assessment Data ---")
    print(f"Generated Equipment Data: {raw_data}")
    print(f"Validation Results: {valid_res}")
    print(f"Diagnostic Results: {diag_res}")
    print("Execution Log:", execution_log)
    print(f"Final Output: System check complete. Average Valid Reading = {avg_val:.2f}\n")

run_exercise_one()
```

```python
LAST_NAME = "ONGJOCO"
SEED_NUM = 2
FAVORITE_ARTIST = "CUP OF JOE"

recursive_call_count = 0
recursive_trace_log = []

def generate_fault_code():
    # Unique numeric fault code derived from credentials
    return len(FAVORITE_ARTIST) * 100 + SEED_NUM * 10 + len(LAST_NAME)

def trace_fault_recursive(current_code, target_condition=0):
    global recursive_call_count
    recursive_call_count += 1
    recursive_trace_log.append(f"Step {recursive_call_count}: Analyzing fault code {current_code}")
    
    # Base condition
    if current_code <= target_condition:
        return "TERMINATED (Base Condition Reached)"
    
    # Recursive reduction step
    next_code = current_code - (SEED_NUM + 5)
    return trace_fault_recursive(next_code, target_condition)

def run_exercise_two():
    print("=== EXERCISE 2: RECURSIVE FAULT TRACE ===")
    fault_code = generate_fault_code()
    execution_log = ["Initialization successful."]
    
    final_res = trace_fault_recursive(fault_code, target_condition=5)
    execution_log.append(f"Recursion completed with {recursive_call_count} calls.")
    
    print("\n--- Assessment Data ---")
    print(f"Generated Fault Data: {fault_code}")
    print("Recursive Trace:", recursive_trace_log)
    print(f"Number of Recursive Calls: {recursive_call_count}")
    print("Execution Log:", execution_log)
    print(f"Final Output: {final_res}\n")

run_exercise_two()
```

```python
import functools

LAST_NAME = "ONGJOCO"
SEED_NUM = 2
FAVORITE_ARTIST = "CUP OF JOE"

execution_log_ex3 = []

# Decorator to monitor major processing function
def monitoring_decorator(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        execution_log_ex3.append(f"[MONITOR] Starting execution of {func.__name__}")
        res = func(*args, **kwargs)
        execution_log_ex3.append(f"[MONITOR] Completed execution of {func.__name__}")
        return res
    return wrapper

# Generator to produce telemetry stream without storing everything at once
def telemetry_generator():
    base = len(LAST_NAME) * 5 + SEED_NUM
    stream = [base, base + 10, 'ERR_STREAM', base + 45, base - 5, 'FAULT_CODE', base + 80]
    for reading in stream:
        yield reading

# Recursive trace function for abnormal conditions (Requirement 7)
def recursive_anomaly_trace(value, threshold=50):
    if value <= threshold:
        return f"Stable ({value})"
    return f"Cascading Warning -> " + recursive_anomaly_trace(value - 20, threshold)

@monitoring_decorator
def process_pipeline():
    valid_results = []
    processed_results = []
    recursive_analysis = []
    valid_count = 0
    invalid_count = 0
    abnormal_count = 0
    
    # Lambda function to filter or transform data
    transform_val = lambda x: x * 1.1 if isinstance(x, (int, float)) else x

    for item in telemetry_generator():
        try:
            # Exception handling for unexpected telemetry types
            if not isinstance(item, (int, float)):
                raise TypeError(f"Corrupted telemetry package: {item}")
            
            val = float(item)
            valid_count += 1
            valid_results.append(f"{val} -> VALID")
            
            transformed = transform_val(val)
            processed_results.append(f"{val} transformed to {transformed:.1f}")
            
            if transformed > 60:
                abnormal_count += 1
                trace_res = recursive_anomaly_trace(int(transformed), threshold=60)
                recursive_analysis.append(f"Value {int(transformed)}: {trace_res}")
                
        except TypeError as te:
            invalid_count += 1
            execution_log_ex3.append(f"EXCEPT: {te}")
            valid_results.append(f"{item} -> INVALID")
            
    total_processed = valid_count + invalid_count
    overall_status = "CRITICAL ALERT" if abnormal_count > 1 else "NOMINAL OPERATION"
    
    return {
        "valid_results": valid_results,
        "processed_results": processed_results,
        "recursive_analysis": recursive_analysis,
        "total_processed": total_processed,
        "valid_count": valid_count,
        "invalid_count": invalid_count,
        "abnormal_count": abnormal_count,
        "status": overall_status
    }

def run_exercise_three():
    print("=== EXERCISE 3: INTELLIGENT EQUIPMENT MONITORING PIPELINE ===")
    data_summary = process_pipeline()
    
    print("\n--- Assessment Data ---")
    print(f"Student-Specific Inputs: LAST_NAME={LAST_NAME}, SEED_NUM={SEED_NUM}, FAVORITE_ARTIST={FAVORITE_ARTIST}")
    print(f"Generated Telemetry Data: Stream via Generator (Artist: {FAVORITE_ARTIST})")
    print(f"Valid/Invalid Results: {data_summary['valid_results']}")
    print(f"Processed Results: {data_summary['processed_results']}")
    print(f"Recursive Analysis: {data_summary['recursive_analysis']}")
    print(f"Final Diagnostic Summary: Total Processed={data_summary['total_processed']}, Valid={data_summary['valid_count']}, Invalid={data_summary['invalid_count']}, Status={data_summary['status']}")
    print("Execution Log:", execution_log_ex3)
    print("Final Output: PIPELINE EXECUTION SUCCESSFUL - REPORT GENERATED")

run_exercise_three()
```
